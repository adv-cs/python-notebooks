## Executing Python code

## Data Types

Python has many of the same fundamental data types as other programming languages.

Like in JS, we can save values into variables using the `=` assignment operator. We don't need to specify the type of variables when creating them (which is required by some languages, like C).

It's best to give your variables **self-documenting** (self-explanatory) names to make your code easy to understand without writing lots of comments.

````python
num_tables = 8
room_temp = 71.4
room_name = "U-212A"
has_windows = True
````

````python
print("Welcome to room", room_name, ". It has", num_tables, "tables and is", room_temp, "°.")
````

When writing output that combines multiple values/variables, it's often easier to use an [**f-string**](https://www.w3schools.com/python/python_string_formatting.asp), which allows for more precise control over formatting.

````python
print(f"Welcome to room {room_name}. It has {num_tables} tables and is {room_temp}°.")
````

We can use the `type()` function to check a variable's type:

````python
print(type(num_tables), type(room_temp), type(room_name), type(has_windows))
````

## Lists

Lists are another data type - specifically, they are an example of a **data structure** because they are used to store other data.

Lists are virtually identical to Arrays in JS, and use similar syntax. They can store essentially any type of data (including lists or other data structure), and can have mixed data types.

````python
schools = ["Trinity", "Riverdale", "Dalton", "Spence"]
temperatures = [48.2, 100, 53, 62]
blue_color_codes = ["blue", "#0000ff", [0, 0, 255]]
empty_list = []
````

We can use the `len()` function to see the size of a list. This also works with other data structures, like strings.

````python
print(len(schools))
print(len(blue_color_codes))
print(len("Trinity")) # number of characters
````

Python lists have many similar methods to similar structures in other languages. For example, we can use the `append` method to add on to lists, just like in JS.

````python
schools.append("Horace Mann")
print(schools)
````

## Operations and Type Conversion

Just like most other programming languages, Python has special operator symbols that allow us to perform operations on data. Different data types support different operations: for example, we can use + with numbers (5 + 2) and strings ("Hello" + "world"), but it will perform different behaviors.

### Arithmetic Operators

Arithmetic operators take numbers (int or float) as operands, and evaluate to an int or float.

````python
print(5 + 2) # addition
print(5 * 2) # multiplication
print(5 ** 2) # exponentiation
````

Two non-obvious arithmetic operations that are implemented in most languages are **integer division** and **modulo** (remainder).

````python
print(10 // 3, 10 % 3)
````

They have a lot of interesting applications. For example, a number *n* is even if, and only if, *n* mod 2 is 0 (in other word, if it's divisible by 2).

````python
n = 5
print(n % 2)
````

Logical operators take boolean (True/False) operands, and return a boolean value. Python's are a little easier to write than those in most other languages (usually `||`, `&&`, and `!`):

### Operations on Sequences

Sequence data types, including **strings** and **lists**, have specific operations.

The **membership operator** `in` lets us check if a value is contained within a particular sequence or data structure; it always evaluates to a boolean.

````python
print("blue" in blue_color_codes)
print("#ff0000" in blue_color_codes)
````

````python
print("i" in "team")
print("yth" in "python")
print("tyh" in "python") # these letters are in "python", but not in the same order (not a substring)
````

The `+` and `*` symbols work on sequences, but not in the same way as numbers!

````python
print("hello" + "goodbye")
print("1" + "1")
print(schools + blue_color_codes)
````

````python
print("10" * 10)
print("-" * 100)
print(["x", "y"] * 5)
````

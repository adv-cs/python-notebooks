There are lots of different data structures; Python has a few language-specific ones built in.

## Set

* Unordered
* No repeated elements
* Mutable
* Members must be immutable elements

Sets allow for much faster search/lookup (`in` operator) than lists.

In practice, sets are useful when we want to be able to check membership in a large collection very quickly (for example: is a word in our 300k English word list?)

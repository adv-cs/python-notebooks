# Sequences and For Loops

## `for` loops

For loops are an area where Python substantially deviates from other languages.

C-family languages like JavaScript tend to favor **count-based loops.** Consider how we might iterate over an array in JS:

```js
animals = ["fish", "lion", "tiger", "bear"]
for (let i = 0; i < animals.length; i++) {
    console.log(animals[i].toUpperCase());
}
```

With help from the Python Toolbox, let's try iterating over a sequence with Python:

````python
animals = ["fish", "lion", "tiger", "bear"]
for animal in animals:
    print(animal.upper())
````

## Using `range` with for loops

While avoiding count-based for loops makes iterating over lists easier, there are times when count-based loops are very useful. Consider this for loop, which prints the even numbers below 20:

```js
for (i = 0; i < 20; i += 2) {
    console.log(i);
}
```

We can replicate this in Python using **`range(start, stop, step)`** (see Toolbox). Note that each parameter of range() matches the order of the original for loop.

````python
for i in range(0, 20, 2):
    print(i)
````

````python
for i in range(10):
    print(i)
````

````python
for i in range(30, 0, -5):
    print(i)
````

**Practice:** Try converting this JS code to Python.

```js
function factorial(n) {
    result = 1;
    
    for (i = n; i > 0; i--) {
        result *= i;
    }

    return result;
}

console.log("3! = " + factorial(3));
console.log("5! = " + factorial(5));
console.log("20! = " + factorial(20));
```

````python
def factorial(n):
    result = 1

    for i in range(n, 0, -1):
        result *= i
    return result

print("3!", factorial(3))
print("5!", factorial(5))
print("20!", factorial(20))
````

## Combining index- and value-based iteration

Sometimes, we might need to keep track of both the index and value when iterating over a sequence. For example, consider this code, which prints out each step in a to-do list with its index.

```js
todo = ["Go shopping", "Do homework", "Sleep"]

for (let i = 0; i < todo.length; i++) {
    console.log((i + 1) + ". " + todo[i])
}
```

Implement the same loop in Python by creating a range that **stops at the length of `todo`**: `range(len(todo))`

````python
todo = ["Go shopping", "Do homework", "Sleep"]

for i in range(len(todo)):
    print(f"{i+1}. {todo[i]}")
````

## What is `range`, really?

Iterating over a `range` is still iterating over a sequence: `range()` creates a special sequence type that we can't access directly:

````python
range(10)
````

````python
type(range(10))
````

We can convert a range to a list to see that it's really just a sequence:

````python
list(range(10))
````
